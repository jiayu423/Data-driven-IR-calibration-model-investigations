```python
def regularized_lvs(lvs, test_score_mean):
    threshold = np.mean(test_score_mean[-7:])
    for i, score_ in enumerate(test_score_mean): 
        if score_ <= threshold: 
            return lvs[i]
```

```python
rgr = PCRegression(4)

rgr.fit(ir.train_ir, ir.train_lc)
pred = rgr.predict(ir.test_ir)
```

```python
s, e = 0, ir.test_dataLList[0]
test_i = [1, 4, 6]
r2_, mae_ = [], []

for i in range(len(ir.test_dataLList)): 

    x, y = ir.test_lc[s:e], pred[s:e]
    r2_.append(r2_score(x, y))
    mae_.append(mean_absolute_error(x, y))

    plt.scatter(x, y, c=clist[test_i[i]])

    s = e
    if i == len(ir.test_dataLList)-1: 
        e = -1
    else: 
        e = e + ir.test_dataLList[i+1]
# plt.legend()
mae_ = np.array(mae_)
plt.title(label=f'R squre={str(np.mean(r2_))[:5]}$\pm${str(np.std(r2_))[:5]} \n MAE={str(np.mean(mae_))[:5]}$\pm${str(np.std(mae_))[:5]}e^-3')
plt.xlabel('y_obs')
plt.ylabel('y_pred (training with 5 exp)')
```

```python
small_train_x = []
small_train_y = []
ind = 1

s, e = 0, ir.train_dataLList[0]

for i in range(len(ir.train_dataLList)): 
    if i == ind: 
        small_train_x.append(ir.train_ir[s+2:e, :])
        small_train_y.append(ir.train_lc[s+2:e, :])
    else: 
        small_train_x.append(ir.train_ir[s:e, :][-1])
        small_train_y.append(ir.train_lc[s:e, :][-1])

    s = e
    if i == len(ir.train_dataLList)-1: 
        e = -1
    else: 
        e = e + ir.train_dataLList[i+1]
        
small_train_x = np.vstack(small_train_x)
small_train_y = np.vstack(small_train_y)
```

```python
small_train_x.shape
```

```python
pca = PCA(20).fit(small_train_x)
plt.scatter([i+1 for i in range(20)], np.cumsum(pca.explained_variance_ratio_))
plt.axhline(y=0.99, c='k', ls='--')
plt.xlabel('PC')
plt.ylabel('explained variance ratio')
plt.title('training on 1 Exp + 3 end points')
```

```python
rgr = PCRegression(3)
rgr.fit(small_train_x,small_train_y)
pred = rgr.predict(ir.test_ir)

s, e = 0, ir.test_dataLList[0]
test_i = [1, 4, 6]
r2_, mae_ = [], []

for i in range(len(ir.test_dataLList)): 

    x, y = ir.test_lc[s:e], pred[s:e]
    r2_.append(r2_score(x, y))
    mae_.append(mean_absolute_error(x, y))

    plt.scatter(x, y, c=clist[test_i[i]])

    s = e
    if i == len(ir.test_dataLList)-1: 
        e = -1
    else: 
        e = e + ir.test_dataLList[i+1]
# plt.legend()
mae_ = np.array(mae_)
plt.title(label=f'R squre={str(np.mean(r2_))[:5]}$\pm${str(np.std(r2_))[:5]} \n MAE={str(np.mean(mae_))[:5]}$\pm${str(np.std(mae_))[:5]}e^-3')
plt.xlabel('y_obs')
plt.ylabel('y_pred (training with 3rd exp+ 4endpoints)')
```
